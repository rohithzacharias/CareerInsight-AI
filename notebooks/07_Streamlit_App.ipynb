```python
import pandas as pd
import joblib
```

```python
model = joblib.load(
    "../models/placement_prediction_model.pkl"
)

print("Model loaded successfully!")
```

```python
def predict_placement(
    gender,
    age,
    degree,
    branch,
    cgpa,
    backlogs,
    internships,
    certifications,
    coding_skills,
    communication_skills,
    aptitude_score,
    projects
):

    overall_skill_score = (
        coding_skills +
        communication_skills +
        aptitude_score
    ) / 3

    academic_performance_index = (
        cgpa * 10
        - backlogs * 5
    )

    profile_strength = (
        internships +
        certifications +
        projects
    )

    student = pd.DataFrame({
        "gender": [gender],
        "age": [age],
        "degree": [degree],
        "branch": [branch],
        "cgpa": [cgpa],
        "backlogs": [backlogs],
        "internships": [internships],
        "certifications": [certifications],
        "coding_skills": [coding_skills],
        "communication_skills": [communication_skills],
        "aptitude_score": [aptitude_score],
        "projects": [projects],
        "overall_skill_score": [overall_skill_score],
        "academic_performance_index": [academic_performance_index],
        "profile_strength": [profile_strength]
    })

    prediction = model.predict(student)[0]

    probability = model.predict_proba(student)[0][1]

    result = "Placed" if prediction == 1 else "Not Placed"

    return result, probability
```

```python
result, probability = predict_placement(
    gender="Male",
    age=21,
    degree="BTech",
    branch="CS",
    cgpa=8.2,
    backlogs=0,
    internships=2,
    certifications=3,
    coding_skills=8,
    communication_skills=8,
    aptitude_score=82,
    projects=3
)

print("Prediction:", result)
print(f"Placement Probability: {probability * 100:.2f}%")
```
